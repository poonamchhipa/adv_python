```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

#Bar Chart – Compare Categories

**Scenario: Compare sales of different products.**

```python
products = ['A', 'B', 'C', 'D']
sales = [100, 200, 150, 180]

plt.bar(products, sales, color='skyblue')
plt.title("Product Sales Comparison")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

# 2. Line Chart – Show Trend Over Time

**Scenario: Website visitors per day,
Use When: You want to show increase/decrease with time.**

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
visitors = [100, 120, 150, 130, 200, 180, 250]

plt.plot(days, visitors, marker='o', color='orange')
plt.title("Website Visitors Over a Week")
plt.xlabel("Days")
plt.ylabel("Visitors")
plt.show()
```

# 3. Histogram – See Data Distribution

**Scenario: Age distribution of employees,
Use When: You want to see how values are spread.**

```python
ages = [22, 25, 27, 30, 31, 28, 24, 26, 29, 40, 42, 38, 35, 32, 36]

plt.hist(ages, bins=5, color='green', edgecolor='black')
plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

# 4. Box Plot – Find Outliers

**Scenario: Salary range and outliers,
Use When: You want to see median, quartiles, and outliers.**

```python
salaries = [25, 28, 30, 32, 35, 40, 45, 60, 70, 100]
sns.boxplot(x=salaries, color='lightblue')
plt.title("Salary Distribution (Box Plot)")
plt.show()
```

# 5. Scatter Plot – Relationship Between Two Variables

**Scenario: Hours studied vs marks scored,
Use When: You want to check correlation between two numeric variables.**

```python
hours = [1, 2, 3, 4, 5, 6, 7]
marks = [40, 45, 55, 60, 70, 75, 85]

plt.scatter(hours, marks, color='red')
plt.title("Hours Studied vs Marks Scored")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.show()
```

# 6. Pie Chart – Show Proportions

**Scenario: Percentage of expenses,
Use When: You want to show part-to-whole relationships.**

```python
labels = ['Rent', 'Food', 'Travel', 'Others']
sizes = [40, 30, 20, 10]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Monthly Expense Breakdown")
plt.show()
```

# 7. Heatmap – Show Correlation Between Many Variables

**Scenario: Correlation between numeric columns.,
Use When: You want to see relationships between many numeric columns.**

```python
data = pd.DataFrame({
    'Age': [22, 25, 28, 30, 32, 35, 40],
    'Salary': [25, 30, 35, 40, 45, 55, 60],
    'Experience': [1, 3, 4, 5, 6, 8, 10]
})

sns.heatmap(data.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

# 8. Scatter Plot with Color (Clusters) – Pattern/Grouping

**Scenario: Visualizing customer clusters,
Use When: You want to visualize clusters or groups in your data.**

```python
df = pd.DataFrame({
    'Income': [30, 35, 40, 80, 85, 90, 25, 28, 95],
    'Spending_Score': [60, 65, 70, 20, 25, 30, 80, 85, 15],
    'Cluster': [0,0,0,1,1,1,2,2,1]
})

sns.scatterplot(x='Income', y='Spending_Score', hue='Cluster', data=df, palette='Set2')
plt.title("Customer Segmentation (Clusters)")
plt.show()
```

#9. Horizontal Bar Chart – Ranking or Top Items

**Scenario: Top 5 movies by rating,
Use When: You want to show ranking or top items.**

```python
movies = ['Movie A', 'Movie B', 'Movie C', 'Movie D', 'Movie E']
ratings = [9.1, 8.5, 8.2, 7.9, 7.5]

plt.barh(movies, ratings, color='purple')
plt.title("Top 5 Movies by Rating")
plt.xlabel("Rating")
plt.ylabel("Movies")
plt.show()
```
